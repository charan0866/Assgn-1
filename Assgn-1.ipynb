```python
def ticketprice(age):

    if (age<4.9):
        print("No need to pay")
    elif(age>4.9 and age<17.1):
        print("Your ticket price is Rs.100")
    elif(age>17.9 and age<60):
        print("Your ticket price is Rs.200")
    else:
        print("Your ticket price is Rs.120")
age = int(input("enter your age: "))

ticketprice(age)

```

```python
def powerbill(units):

    if(units<101):
        return units*3
    elif(units>100 and units<201):
        return(300+((units-100)*5))
    else:
        return(800+((units-200)*8))
units = int(input("Enter the number of units consumed: "))

print("The total is ", powerbill(units))
```

```python
marks = int(input("Enter your marks"))

if (marks<101 and marks>89):
    print("your grade is 'A' ")
elif (marks<90 and marks>74):
    print("your grade is 'B' ")
elif (marks<75 and marks>49):
    print("your grade is 'C' ")
elif (marks<50 and marks>34):
    print("your grade is 'D' ")
elif (marks<35 and marks>-1):
    print("You're Failed")
else:
    print("Invalid marks")

```

```python
billamt = int(input("Enter the bill value"))
val = 10
sofdig = 0
while(billamt>0):

    dig = billamt%val
    sofdig = sofdig+dig
    billamt = billamt//10

print("Sum of digits is",sofdig)

if (sofdig>20):
    print("Congratulations! you win a free gift")
else:
    print("Better luck next time")


```
