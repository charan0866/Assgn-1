```python
sen = input("Enter a sentence: ")
sen = sen.lower()
v= ['a','e','i','o','u']

vo = 0
co = 0
sp = 0

for letter in sen:
    if letter in v:
        vo += 1
    
    elif letter == " ":
        sp += 1
    else:
        co+= 1

print("Vowels:", vo)
print("Consonants:", co)
print("Spaces:", sp)
```

```python
temps = [32, 35, 38, 41, 36, 29, 40]
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

hightemp = max(temps)


for i in range(len(temps)):
    if (temps[i] > 37):
        print("Heat alert days:",days[i], "-", temps[i], "°C")
for j in range(len(temps)):
    if (hightemp  == temps[j]):
        print("The hottest day is",days[j])
f = []
for t in temps:
    ft = t * 9 / 5 + 32
    f.append(int(ft))

print("Temperatures in Fahrenheit:", f)

```

```python
sent = input("Enter a sentence: ")
words = sent.split()

lngst = words[0]
shrtst = words[0]
c = 0

for i in words:
    if len(i) > len(lngst):
        lngst = i
    if len(i) < len(shrtst):
        shrtst = i
    if i[0].isupper():
        c += 1

print("Longest word:", lngst)
print("Shortest word:", shrtst)
print("Words starting with a capital letter:", c)
```

```python
balances = [1500, -200, 3000, -50, 0, 750, -1000]

posacc = []
negacc = []

for i in balances:
    if i > 0:
        posacc.append(i)
    elif i < 0:
        negacc.append(i)

totl = 0
for i in negacc:
    totl += i

print("Positive accounts:", posacc)
print("Negative accounts:", negacc)
print("Total amount owed:", totl)
```

```python
pswd = input("Enter a password: ")


scl=[]

for ch in pswd:
    if ch.isupper():
        scl.append('1')

    elif ch.islower():
        scl.append('2')
    elif ch.isdigit():
        scl.append('3')

leng = len(pswd)


if (leng>7):
    scl.append('4')

if len(set(scl)) == 4:
    print("Strong")
elif len(set(scl)) > 1:
    print("Medium")
else:
    print("Weak")
```

```python
name = input("Enter full name: ")
wds = name.split()

ini = ""
for i in wds:
    ini += i[0].upper() + "."

titlename = ""
for i in wds:
    titlename += i.capitalize() + " "


idname = ""
for i in wds[:-1]:
    idname += i[0].upper() + ". "
idname += wds[-1].capitalize()

print("Initials:", ini)
print("Title case:", titlename)
print("ID card name:", idname)
```
