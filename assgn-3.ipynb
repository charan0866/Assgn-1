```python
contacts = {}

while True:
    print("\n--- Phone Contact Book ---")
    print("1. Add contact")
    print("2. Search contact")
    print("3. Show all contacts")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        name = input("Enter contact name: ")
        phone = input("Enter phone number: ")

        if name in contacts:
            print("Contact already exists")
        else:
            contacts[name] = phone
            print("Contact added successfully")

    elif choice == "2":
        name = input("Enter name to search: ")

        if name in contacts:
            print("Phone number:", contacts[name])
        else:
            print("Contact not found")

    elif choice == "3":
        
        print("\nContacts:")
        for name, phone in contacts.items():
            print(name, ":", phone)

    elif choice == "4":
        print("Exiting Contact Book...")
        break

    else:
        print("Invalid choice")
```

```python
marks = {
    "Aman": 82,
    "Priya": 91,
    "Rahul": 45,
    "Sneha": 67,
    "Kiran": 30
}


for name, score in marks.items():
    if score >= 40:
        print(name, "- Pass")
    else:
        print(name, "- Fail")

top = []

for m in marks:
    top.append(marks[m])


topper = max(top)

for n,m in marks.items():
    if (m == topper):
        print("Topper is",n)

print("Topper marks:", topper)

# Create dictionary containing only passed students
passed_students = {}

for name, score in marks.items():
    if score >= 40:
        passed_students[name] = score

print("\nPassed students:")
print(passed_students)
```

```python
riya_friends = {"Aman", "Neha", "Karan", "Sita", "Vikram"}

arjun_friends = {"Neha", "Vikram", "Pooja", "Aman", "Rohan"}


mutual_friends = riya_friends & arjun_friends

only_riya = riya_friends - arjun_friends

all_friends = riya_friends | arjun_friends

suggestions = arjun_friends - riya_friends

print("Mutual friends:", mutual_friends)
print("Friends only Riya has:", only_riya)
print("All people:", all_friends)
print("Friend suggestions for Riya:", suggestions)
```

```python
text = input("Enter text: ")

character_count = {}

for char in text:
    if char != " ":
        if char in character_count:
            character_count[char] += 1
        else:
            character_count[char] = 1

print(character_count)

if character_count:
    most_frequent = max(character_count, key=character_count.get)

    print("Most frequent character:",
        most_frequent,
        "(" + str(character_count[most_frequent]) + " times)")
```

```python
menu = {
    "dosa": 60,
    "idli": 40,
    "vada": 30,
    "coffee": 25,
    "tea": 15
}

total = 0

print("Menu:")
for item, price in menu.items():
    print(item, "- Rs.", price)

while True:
    order = input("\nEnter item (or type 'bill'): ")

    if order == "bill":
        break

    if order in menu:
        total += menu[order]
        print(order, "added")
    else:
        print("Item not available")

print("\nBill before discount: Rs.", total)

if total > 200:
    discount = total * 0.1
    total = total - discount

    print("10% discount: Rs.", discount)

print("Total bill: Rs.", total)
```

```python
books = {
    "Python Basics": 3,
    "Data Science 101": 0,
    "Machine Learning": 2,
    "SQL Guide": 1
}

book_name = input("Enter book name to borrow: ")

if book_name not in books:
    print("Book not found")

elif books[book_name] == 0:
    print("Currently unavailable")

else:
    books[book_name] -= 1
    print("Book issued successfully")

print("\nUpdated book availability:")
print(books)
```

```python
employees = [
    ("Ravi", "IT"),
    ("Anjali", "HR"),
    ("Suresh", "IT"),
    ("Meena", "Finance"),
    ("Arun", "HR"),
    ("Divya", "IT")
]

departments = {}

for name, department in employees:

    if department not in departments:
        departments[department] = []

    departments[department].append(name)

print("Employees grouped by department:")
print(departments)

print("\nEmployee count:")

for department, enames in departments.items():
    print(department, ":", len(enames))
```

```python
answer_key = ("B", "C", "A", "D", "B")

student_answers = []


for i in range(5):
    answer = input(
        "Enter answer for question " + str(i + 1) + ": "
    ).upper()

    student_answers.append(answer)

wrong_questions = set()
correct_questions = []
score = 0

for i in range(5):

    if student_answers[i] == answer_key[i]:
        correct_questions.append(i + 1)
        score += 1
    else:
        wrong_questions.add(i + 1)

print("\nCorrect question numbers:", correct_questions)
print("Wrong question numbers:", wrong_questions)

print("Final score:", score, "out of 5")

if score == 5:
    print("Remark: Excellent")

elif score >= 3:
    print("Remark: Good")

else:
    print("Remark: Needs improvement")
```
